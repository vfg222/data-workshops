# Seoul Bike Sharing Demand (W14)

This week we will work with the Seoul Bike Sharing Demand dataset from the [UCI Machine Learning Repository](https://archive.ics.uci.edu/dataset/560/seoul+bike+sharing+demand) ([DOI](https://doi.org/10.24432/C5F62R)). The `SeoulBikeData.csv` file contains 8,760 hourly observations from the late 2010s. Each row represents one hour in Seoul's public bike system, with the number of bikes rented recorded as `Rented Bike Count` and a mix of interesting features. The objective is to forecast the number of bikes rented on a future day under any specified combination of feature values.

For a bit of real-world context: Seoul's public bicycle-sharing system (often called Seoul Bike or "Ttareungi") was rolled out by the Seoul Metropolitan Government after a trial period and began operating in selected neighborhoods in October 2015. A short official background article from the city can be found [here](https://english.seoul.go.kr/ring-ring-enjoy-riding-seoul-seoul-bike/).

The columns are easy to think about in a few groups:
- time: `Date`, `Hour`, `Seasons`
- target: `Rented Bike Count`
- weather: `Temperature(°C)`, `Humidity(%)`, `Wind speed (m/s)`, `Visibility (10m)`, `Dew point temperature(°C)`, `Solar Radiation (MJ/m2)`, `Rainfall(mm)`, `Snowfall (cm)`
- context: `Holiday`, `Functioning Day`
